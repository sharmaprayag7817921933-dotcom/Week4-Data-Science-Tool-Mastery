```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create a sample dataset
data = {
    "Student": ["Aman", "Priya", "Rahul", "Neha", "Rohan",
                "Simran", "Arjun", "Anjali", "Karan", "Pooja"],
    "Study_Hours": [2, 5, 3, 6, 4, 7, 1, 5, 3, 6],
    "Marks": [45, 85, 55, 90, 65, 95, 35, 80, 50, 88],
    "Attendance": [70, 95, 75, 98, 80, 99, 60, 90, 72, 96]
}

df = pd.DataFrame(data)

print("DATA SCIENCE TOOL MASTERY PROJECT")
print("=" * 40)
print("Dataset created successfully!")

display(df)
```

```python
# STEP 4: DATA EXPLORATION

print("1. DATASET SHAPE")
print("Rows and Columns:", df.shape)

print("\n2. DATASET INFORMATION")
df.info()

print("\n3. STATISTICAL SUMMARY")
display(df.describe())

print("\n4. MISSING VALUES")
print(df.isnull().sum())

print("\n5. DUPLICATE RECORDS")
print(df.duplicated().sum())

print("\n6. COLUMN NAMES")
print(df.columns.tolist())
```

```python
# STEP 5: DATA VISUALIZATION

plt.figure(figsize=(10, 5))

plt.bar(df["Student"], df["Marks"])

plt.title("Student Performance Analysis")
plt.xlabel("Student Name")
plt.ylabel("Marks")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Study_Hours"], df["Marks"])

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.grid(True)
plt.show()
```

```python
# STEP 7: CORRELATION ANALYSIS

correlation = df[["Study_Hours", "Marks", "Attendance"]].corr()

print("CORRELATION ANALYSIS")
display(correlation)

plt.figure(figsize=(7, 5))

plt.imshow(correlation, cmap="coolwarm", vmin=-1, vmax=1)

plt.xticks(range(3), correlation.columns, rotation=45)
plt.yticks(range(3), correlation.columns)

plt.colorbar(label="Correlation")
plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```

```python
# STEP 8: FINAL ANALYSIS AND FINDINGS

print("WEEK 4: DATA SCIENCE TOOL MASTERY PROJECT")
print("=" * 45)

print("\n1. DATASET SUMMARY")
print("Total Students:", len(df))
print("Total Columns:", len(df.columns))

print("\n2. PERFORMANCE ANALYSIS")
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())
print("Average Marks:", round(df["Marks"].mean(), 2))

print("\n3. STUDY ANALYSIS")
print("Average Study Hours:", round(df["Study_Hours"].mean(), 2))

print("\n4. ATTENDANCE ANALYSIS")
print("Average Attendance:", round(df["Attendance"].mean(), 2), "%")

print("\n5. TOP PERFORMER")
top_student = df.loc[df["Marks"].idxmax(), "Student"]
print("Student:", top_student)
print("Marks:", df["Marks"].max())

print("\n6. DATA QUALITY")
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Records:", df.duplicated().sum())

print("\n7. CONCLUSION")
print("Python, Pandas, NumPy, and Matplotlib were used")
print("to explore, analyze, and visualize the dataset.")
print("Graphs and statistics helped summarize the sample data.")
print("\nPROJECT COMPLETED SUCCESSFULLY!")
```
